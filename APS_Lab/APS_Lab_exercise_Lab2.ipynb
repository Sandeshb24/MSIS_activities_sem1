## Addition rule for disjoint sets


```python
import random

total_rolls = 10000

rolls = [random.randint(1, 6) for _ in range(total_rolls)]

count_A = rolls.count(1)
count_B = rolls.count(6)

count_A_and_B = sum(1 for roll in rolls if roll == 1 and roll == 6)

count_A_or_B = sum(1 for roll in rolls if roll == 1 or roll == 6)

prob_A = count_A / total_rolls
prob_B = count_B / total_rolls
prob_A_and_B = count_A_and_B / total_rolls
prob_A_or_B = count_A_or_B / total_rolls

print("--- 1. Event Classification ---")
if prob_A_and_B == 0:
    print("Events A and B are DISJOINT because P(A and B) = 0 (they cannot happen simultaneously).")
else:
    print("Events A and B are JOINT.")

print("\n--- 2. Addition Rule Verification ---")
print(f"P(A) = {prob_A:.4f}")
print(f"P(B) = {prob_B:.4f}")
print(f"P(A) + P(B) = {prob_A + prob_B:.4f}")
print(f"P(A u B)    = {prob_A_or_B:.4f}")

if abs(prob_A_or_B - (prob_A + prob_B)) < 1e-9:
    print("\nVerification Successful: P(A u B) == P(A) + P(B)")
```

Addition rule for overlapping events

For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}

Verify: P(A∪B)=P(A)+P(B)−P(A∩B)

```python
import random

total_rolls = 10000

rolls = [random.randint(1, 6) for _ in range(total_rolls)]

count_A = sum(1 for roll in rolls if roll in [2, 4, 6])
count_B = sum(1 for roll in rolls if roll > 3)

count_A_and_B = sum(1 for roll in rolls if (roll in [2, 4, 6]) and (roll > 3))

count_A_or_B = sum(1 for roll in rolls if (roll in [2, 4, 6]) or (roll > 3))

prob_A = count_A / total_rolls
prob_B = count_B / total_rolls
prob_A_and_B = count_A_and_B / total_rolls
prob_A_or_B = count_A_or_B / total_rolls

print("--- 1. Event Classification ---")
print(f"P(A and B) = {prob_A_and_B:.4f}")
if prob_A_and_B > 0:
    print("Events A and B are OVERLAPPING (Joint) because P(A and B) > 0.")

print("\n--- 2. Addition Rule Verification ---")
print(f"P(A) = {prob_A:.4f}")
print(f"P(B) = {prob_B:.4f}")
print(f"P(A and B) = {prob_A_and_B:.4f}")

calculated_union = prob_A + prob_B - prob_A_and_B

print(f"P(A) + P(B) - P(A and B) = {calculated_union:.4f}")
print(f"P(A u B)                 = {prob_A_or_B:.4f}")

if abs(prob_A_or_B - calculated_union) < 1e-9:
    print("\nVerification Successful: P(A u B) == P(A) + P(B) - P(A and B)")
```

Multiplication rule for independent events

A coin is tossed and a die is rolled 10,000 times.

Define:

obtaining Head  

obtaining an even number

Verify: P(A∩B)=P(A)P(B)

```python
import random

total_trials = 10000

coin_tosses = [random.choice(['H', 'T']) for _ in range(total_trials)]
die_rolls = [random.randint(1, 6) for _ in range(total_trials)]

count_A = coin_tosses.count('H')
count_B = sum(1 for roll in die_rolls if roll in [2, 4, 6])

count_A_and_B = sum(
    1 for toss, roll in zip(coin_tosses, die_rolls)
    if toss == 'H' and roll in [2, 4, 6]
)

prob_A = count_A / total_trials
prob_B = count_B / total_trials
prob_A_and_B = count_A_and_B / total_trials

calculated_intersection = prob_A * prob_B

print("--- Multiplication Rule Verification ---")
print(f"P(A) [Head]            = {prob_A:.4f}")
print(f"P(B) [Even Roll]       = {prob_B:.4f}")
print(f"P(A) * P(B)            = {calculated_intersection:.4f}")
print(f"P(A and B) [Observed]  = {prob_A_and_B:.4f}")

if abs(prob_A_and_B - calculated_intersection) < 1e-3:
    print("\nVerification Successful: P(A and B) \u2248 P(A) * P(B)")
```
