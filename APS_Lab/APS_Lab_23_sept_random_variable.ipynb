Problem set 1: Simulating Customer Clicks Using a Bernoulli Random Variable




```python
import numpy as np

p = 0.3
sample_size = 10000
n = 1


```

```python
samples= np.random.binomial(n = 1, p =p,size = sample_size)
```

```python
print("outome for 20 customers are:")
print(samples)
```

```python
#calculating the  probabilites of 1s and 0s
count_0,count_1 = np.bincount(samples)
```

```python
print(f"p(X=1) =  {(count_1/sample_size)*100}%")
print(f"p(X=0) =  {(count_0/sample_size)*100}%")
```

```python
number_of_clicks = np.sum(samples == 1)

number_of_non_clicks = np.sum(samples == 0)

print("Number of clicks:", number_of_clicks)
print("Number of non-clicks:", number_of_non_clicks)
```

```python
import matplotlib.pyplot as plt
```

```python
outcomes = ['0 (Non-click)', '1 (Click)']
x = np.arange(len(outcomes))
width = 0.35

# 2. Theoretical vs Practical PMF values
theoretical_pmf = [1 - p, p]
practical_pmf = [count_0 / sample_size, count_1 / sample_size]

# 3. Plot side-by-side bars
plt.figure(figsize=(7, 5))
plt.bar(x - width/2, theoretical_pmf, width, label='Theoretical PMF', color='royalblue')
plt.bar(x + width/2, practical_pmf, width, label='Practical PMF', color='orange')

# 4. Customization
plt.xlabel('Customer Outcome')
plt.ylabel('Probability')
plt.title('Theoretical vs Practical PMF')
plt.xticks(x, outcomes)
plt.ylim(0, 1)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Display plot
plt.show()
```

```python

expectation_theoretical = 1*p + 0*(1-p)
```

```python
expectation_imperical = np.mean(samples)
print("Theoretical and empircal expectation are:",expectation_theoretical,expectation_imperical)
```

```python
var_theo = pow(1,2)+(pow(0,2)*(1-p))- pow(expectation_theoretical,2)
```

```python
print("Theoretical Variance:", var_theo)
print("Empirical Variance:", var_emp)
```
