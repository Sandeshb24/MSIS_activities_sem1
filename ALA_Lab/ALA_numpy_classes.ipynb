```python
v = np.array([1.5, 2.0, 3.0])
```

```python
W = np.array([
    [0.2,  0.8],
    [-0.5, 0.1],
    [0.8, -0.3]
])


b = np.array([0.5, -0.2])

# Activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
z = np.dot(v, W) + b

# 2. Apply activation
a = sigmoid(z)

print("Hidden layer activations:", a)
```
