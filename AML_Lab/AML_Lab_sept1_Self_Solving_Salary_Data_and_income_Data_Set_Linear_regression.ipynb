```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/Salary_Data.csv")
```

```python
df.head()
```

```python
df.describe()
```

```python
X  = df.iloc[:,[0]]
```

```python
X.head()
```

```python
y = df.iloc[:,[1]]
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_train, x_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state= 0 )
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
reg.fit(X_train,y_train)
```

```python
y_pred = reg.predict(x_test)
```

```python
print("Intercept",reg.intercept_)
print("Slope", reg.coef_[0])
print("Actual values",y_test)
print("Predicted values",y_pred)
```

```python
from sklearn.metrics import mean_absolute_error
print("mean absolute error:",mean_absolute_error(y_test,y_pred))
print("mean squared error:", mean_squared_error(y_test,y_pred))
print("root mean squared error:",np.sqrt(mean_squared_error(y_test,y_pred)))
print("r2 score:",r2_score(y_test,y_pred))
```

```python
import matplotlib.pyplot as plt
plt.scatter(y_test,y_pred, color = "blue")

minimum = min(y_test.min().item(), y_pred.min())
maximum = max(y_test.max().item(), y_pred.max())
plt.plot([minimum, maximum], [minimum, maximum], color = "red", linestyle = "--")

plt.xlabel("Actual Salary")
plt.ylabel("Predicted Salary")
plt.title("Actual vs Predicted Salary")
plt.show()
```

```python
#residual plot
residuals = y_test - y_pred
plt.scatter(y_pred,residuals,color = "purple")
plt.axhline(y = 0, color = "red", linestyle = "--")
plt.xlabel("Predicted Marks")
plt.ylabel("Residuals")
plt.title("Residual Plot")
plt.show()
```

# Income_data set linear reg

```python
df2 = pd.read_csv("/content/income_data.csv")
```

```python
df2.head()
```

```python
X = df2[["income"]]
y = df2["happiness"]
```

```python
model = LinearRegression()
```

```python
X_train, x_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state= 0 )
```

```python
model.fit(X_train,y_train)
```

```python
y_pred_income = model.predict(x_test)
```

```python
print("Intercept", model.intercept_)
print("Slope", model.coef_[0])
print("Actual values", y_test.head())
print("Predicted values", y_pred_income[:5])
```

```python
print("mean absolute error:", mean_absolute_error(y_test, y_pred_income))
print("mean squared error:", mean_squared_error(y_test, y_pred_income))
print("root mean squared error:", np.sqrt(mean_squared_error(y_test, y_pred_income)))
print("r2 score:", r2_score(y_test, y_pred_income))
```

```python
import matplotlib.pyplot as plt
plt.scatter(y_test, y_pred_income, color = "blue")

minimum_income = min(y_test.min(), y_pred_income.min())
maximum_income = max(y_test.max(), y_pred_income.max())
plt.plot([minimum_income, maximum_income], [minimum_income, maximum_income], color = "red", linestyle = "--")

plt.xlabel("Actual Happiness")
plt.ylabel("Predicted Happiness")
plt.title("Actual vs Predicted Happiness (Income Data)")
plt.show()
```

```python
# Residual plot for income_data
residuals_income = y_test - y_pred_income
plt.scatter(y_pred_income, residuals_income, color = "purple")
plt.axhline(y = 0, color = "red", linestyle = "--")
plt.xlabel("Predicted Happiness")
plt.ylabel("Residuals")
plt.title("Residual Plot (Income Data)")
plt.show()
```
